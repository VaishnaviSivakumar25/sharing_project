```python
plt.figure(figsize=(10,6))
def autopct_with_counts(pct, values):
    total = sum(values)
    absolute = int(pct/100 * total)
    return f'{pct:.1f}%\n ({absolute})'
    
uber['CATEGORY'].value_counts().plot.pie(autopct = lambda pct: autopct_with_counts(pct,uber['CATEGORY'].value_counts()), startangle  = 90, cmap = 'Set3', labels = None )
plt.title('porporation of categories with business slice with Counts')
plt.ylabel('')
plt.legend(labels = uber['CATEGORY'].value_counts().index,loc= 'upper right')
plt.show()
```

## average time difference between the start of the first ride and the start of the next ride of the day?

```python
uber['START_DATE'] = pd.to_datetime(uber['START_DATE'])
```

```python
uber['date'] = uber['START_DATE'].dt.date
```

```python
uber.columns
```

```python
ubber = uber.sort_values('START_DATE')
uber['next_ride'] = uber.groupby('date')['START_DATE'].shift(-1)
uber['time_diff'] = (uber['next_ride'] - uber['START_DATE']).dt.total_seconds() / 60
average_time = uber.groupby('date')['time_diff'].mean().dropna()
average_time
```
